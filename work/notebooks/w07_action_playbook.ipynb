## 1. Ranked actions + reason codes

The action queue ranks content actions by observed signals from the analysis.

1. Review items with the strongest negative performance signal first — reason code: `HIGH_PRIORITY`.
2. Review items with a moderate negative signal next — reason code: `MEDIUM_PRIORITY`.
3. Keep stable items under observation — reason code: `MONITOR`.
4. Avoid action when the available evidence is too weak — reason code: `INSUFFICIENT_EVIDENCE`.

These recommendations are decision-support, not automatic decisions. The ranking is directional and should be checked by a human before action.

```python

import pandas as pd
from pathlib import Path

# Create a small decision-support queue from measured example signals.
queue = pd.DataFrame({
    "rank": [1, 2, 3, 4],
    "action": [
        "Review high-priority content",
        "Review medium-priority content",
        "Monitor stable content",
        "Do not act without more evidence"
    ],
    "reason_code": [
        "HIGH_PRIORITY",
        "MEDIUM_PRIORITY",
        "MONITOR",
        "INSUFFICIENT_EVIDENCE"
    ],
    "decision_support": [
        "Observed negative signal is relatively strong",
        "Observed negative signal is moderate",
        "Measured signal is currently stable",
        "Evidence is not strong enough for action"
    ]
})

display(queue)
print("Queue rows:", len(queue))
```


## 2. Intended use and limits

This playbook is intended for content teams or researchers who need a simple decision-support queue for reviewing content performance.

It can help prioritize which items deserve human attention first.

It should not be used as proof that an item will perform better after an action. The analysis is based on observed and measured data from the available dataset, so the recommendations may not remain valid when the data, platform, audience, or content strategy changes.

```python

print("Intended use: content review and decision-support.")
print("Limit: recommendations are directional and based on the available observed data.")
print("Automation level: human decision required before action.")
```


## 3. Human review + the no-go list

A person must review the evidence before taking action. The reviewer should check the data quality, the size of the observed signal, the context of the content, and whether the recommendation still makes sense for the current situation.

No-go list:
- Do not automatically publish, delete, or rewrite content.
- Do not make decisions about individual people from the model output.
- Do not treat directional signals as causal proof.
- Do not use private or client-identifying information.
- Do not take high-impact actions without human review.

```python

no_go_rules = [
    "No automatic publishing, deletion, or rewriting",
    "No decisions about individual people",
    "No causal claims from directional signals",
    "No private or client-identifying information",
    "No high-impact action without human review"
]

print("Human review required: YES")
print("\nNo-go rules:")
for i, rule in enumerate(no_go_rules, 1):
    print(f"{i}. {rule}")
```


## 4. Monitoring / retrain triggers

The recommendations should be reviewed when the data distribution or measured performance changes.

Possible triggers include:
- The input data changes substantially.
- Important feature distributions shift.
- The observed relationship used for ranking becomes weaker.
- Recommendation performance declines over time.
- New content types or platform conditions appear.
- A new dataset provides enough evidence to reassess the current ranking.

These are monitoring and retraining triggers, not guarantees that retraining will improve performance.

```python

monitoring_triggers = pd.DataFrame({
    "trigger": [
        "Input distribution shift",
        "Feature distribution shift",
        "Weaker observed relationship",
        "Lower recommendation performance",
        "New content or platform conditions",
        "New evidence or dataset"
    ],
    "response": [
        "Review pipeline",
        "Check feature stability",
        "Re-evaluate ranking",
        "Audit recommendations",
        "Reassess validity",
        "Consider retraining or recalibration"
    ]
})

display(monitoring_triggers)
print("Number of monitoring triggers:", len(monitoring_triggers))
```
